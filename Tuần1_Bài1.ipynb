```python
# Import các thư viện cần thiết
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer

# Tải tập dữ liệu Breast Cancer từ Scikit-learn
data = load_breast_cancer()

# Chuyển đổi thành DataFrame để xử lý bằng Pandas
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

# 1. Xem danh sách các thuộc tính của dữ liệu
print("=== DANH SÁCH CÁC THUỘC TÍNH ===")
print(df.columns.tolist())

# 2. Xem kích thước dữ liệu và nhãn
print("\n=== KÍCH THƯỚC DỮ LIỆU VÀ NHÃN ===")
print("Kích thước ma trận đặc trưng (Features):", data.data.shape)
print("Kích thước nhãn (Target):", data.target.shape)
print("Các lớp nhãn:", data.target_names) # 0: malignant (ác tính/bệnh), 1: benign (lành tính)

# 3. Lọc ra những mẫu có bệnh ung thư (Malignant - nhãn 0 trong dataset của sklearn)
cancer_samples = df[df['target'] == 0]
print("\n=== SỐ MẪU BỆNH UNG THƯ (ÁC TÍNH) ===")
print(f"Số lượng mẫu: {len(cancer_samples)}")
display(cancer_samples.head())

# 4. Lọc ra những mẫu lành tính (Benign - nhãn 1 trong dataset của sklearn)
benign_samples = df[df['target'] == 1]
print("\n=== SỐ MẪU LÀNH TÍNH ===")
print(f"Số lượng mẫu: {len(benign_samples)}")
display(benign_samples.head())

# 5. Tính toán độ tương quan của từng thuộc tính với thuộc tính bệnh (target)
correlation = df.corr()['target'].sort_values()
print("\n=== ĐỘ TƯƠNG QUAN CỦA TỪNG THUỘC TÍNH VỚI NHÃN (TARGET) ===")
print(correlation)

# 6. Vẽ biểu đồ thể hiện độ tương quan giữa các thuộc tính (Heatmap)
plt.figure(figsize=(16, 12))
sns.heatmap(df.corr(), cmap='coolwarm', annot=False, linewidths=0.5)
plt.title('Biểu đồ nhiệt tương quan giữa các thuộc tính (Correlation Heatmap)')
plt.show()

# 7. Trực quan hóa tỷ lệ nhãn
plt.figure(figsize=(6, 6))
label_counts = df['target'].value_counts()
labels = [data.target_names[i] for i in label_counts.index]

plt.pie(label_counts, labels=labels, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff'])
plt.title('Tỷ lệ mẫu Bệnh ung thư (Malignant) vs Lành tính (Benign)')
plt.show()
```
