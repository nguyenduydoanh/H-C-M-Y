Bài 6 Chuyển đổi kiểu dữ liệu

```python
# 1. Hàm chuẩn hóa Price_VND
def clean_price(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()

    # "miễn phí" -> 0.0 VND
    if 'miễn phí' in val_str or 'free' in val_str:
        return 0.0

    # Bỏ đơn vị tiền tệ 'đ', 'vnd'
    val_str = val_str.replace('đ', '').replace('vnd', '').strip()

    # Bỏ dấu phân cách nghìn (dấu chấm hoặc phẩy)
    val_str = val_str.replace('.', '').replace(',', '')

    try:
        return float(val_str)
    except ValueError:
        return np.nan

# 2. Hàm chuẩn hóa Rating
def clean_rating(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()
    val_str = val_str.replace('sao', '').replace('stars', '').replace('star', '').strip()
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# 3. Hàm chuẩn hóa Purchase_Count
def clean_purchase_count(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower().replace('lượt', '').strip()

    # Đổi hậu tố 'k' (ví dụ 1.2k -> 1200)
    if 'k' in val_str:
        val_str = val_str.replace('k', '').strip()
        try:
            return float(val_str) * 1000
        except ValueError:
            return np.nan
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# 4. Hàm chuẩn hóa Weight (quy đổi toàn bộ về gram)
def clean_weight(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower().replace('nặng', '').strip()

    if 'kg' in val_str:
        num_part = val_str.replace('kg', '').strip()
        try:
            return float(num_part) * 1000.0  # 1 kg = 1000 g
        except ValueError:
            return np.nan
    elif 'g' in val_str:
        num_part = val_str.replace('g', '').strip()
        try:
            return float(num_part)
        except ValueError:
            return np.nan
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# 5. Hàm chuẩn hóa Discount (quy đổi về dạng thập phân 0 - 1)
def clean_discount(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()

    if 'phần trăm' in val_str:
        num_part = val_str.replace('phần trăm', '').strip()
        try:
            return float(num_part) / 100.0
        except ValueError:
            return np.nan
    if '%' in val_str:
        num_part = val_str.replace('%', '').strip()
        try:
            return float(num_part) / 100.0
        except ValueError:
            return np.nan
    try:
        return float(val_str)
    except ValueError:
        return np.nan

# ==========================================
# THỰC THI CHUYỂN ĐỔI TRÊN DF_CLEAN
# ==========================================
# Tạo lại bản sao mới tinh từ dữ liệu gốc df_raw
df_clean = df_raw.copy()

cols = ['Price_VND', 'Rating', 'Purchase_Count', 'Weight', 'Discount', 'Created_Date']

# 1. Lấy kiểu dữ liệu gốc thực sự từ df_raw
dtypes_before = df_raw[cols].dtypes

# 2. Áp dụng chuyển đổi sang df_clean
df_clean['Price_VND'] = df_clean['Price_VND'].apply(clean_price)
df_clean['Rating'] = df_clean['Rating'].apply(clean_rating)
df_clean['Purchase_Count'] = df_clean['Purchase_Count'].apply(clean_purchase_count)
df_clean['Weight'] = df_clean['Weight'].apply(clean_weight)
df_clean['Discount'] = df_clean['Discount'].apply(clean_discount)
df_clean['Created_Date'] = pd.to_datetime(df_clean['Created_Date'], format='mixed', dayfirst=True, errors='coerce')

# 3. Lấy kiểu dữ liệu sau chuyển đổi từ df_clean
dtypes_after = df_clean[cols].dtypes

# 4. In bảng đối chiếu trước - sau
df_compare = pd.DataFrame({
    'Kiểu ban đầu (df_raw)': dtypes_before.astype(str),
    'Kiểu sau xử lý (df_clean)': dtypes_after.astype(str)
})

print("=== SO SÁNH KIỂU DỮ LIỆU TRƯỚC VÀ SAU ===")
print(df_compare)

print("\n=== THÔNG TIN CẤU TRÚC SAU KHI LÀM SẠCH ===")
df_clean[cols].info()
```

```python
print("=== 1. KIỂM TRA TRƯỚC KHI XỬ LÝ ===")
n_rows_before = len(df_clean)
exact_dups_count = df_clean.duplicated().sum()
id_dups_count = df_clean['Product_ID'].duplicated().sum()

print(f"Tổng số dòng ban đầu: {n_rows_before}")
print(f"Số dòng trùng lặp hoàn toàn: {exact_dups_count}")
print(f"Số dòng trùng khóa Product_ID: {id_dups_count}")

# 2. Xóa các dòng trùng lặp hoàn toàn
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
n_rows_after_exact = len(df_clean)
print(f"\nSố dòng sau khi xóa trùng hoàn toàn: {n_rows_after_exact} (giảm {n_rows_before - n_rows_after_exact} dòng)")

# 3. Nhận diện các Product_ID còn bị xung đột nội dung
conflict_ids = df_clean[df_clean.duplicated(subset=['Product_ID'], keep=False)]
print("\n=== CÁC DÒNG BỊ TRÙNG MÃ PRODUCT_ID NHƯNG KHÁC NỘI DUNG ===")
print(conflict_ids[['Product_ID', 'Product_Name', 'Category', 'Price_VND', 'Seller_City', 'Created_Date']])

# 4. Gắn cờ Duplicate_ID_Flag để phục vụ kiểm toán và không làm mất thông tin
df_clean['Duplicate_ID_Flag'] = df_clean.duplicated(subset=['Product_ID'], keep=False).astype(int)

# 5. Đề xuất quy tắc giữ 1 bản ghi duy nhất nếu bài toán yêu cầu:
# Quy tắc ưu tiên: Giữ dòng có thời gian tạo mới nhất (Created_Date lớn hơn), nếu bằng thì giữ dòng ít NaN hơn
df_clean_unique = df_clean.sort_values(by=['Product_ID', 'Created_Date'], ascending=[True, True])
df_clean_unique = df_clean_unique.drop_duplicates(subset=['Product_ID'], keep='last').reset_index(drop=True)

print(f"\n=== BÁO CÁO TỔNG KẾT DÒNG ===")
print(f"- Ban đầu: {n_rows_before} dòng")
print(f"- Sau khi loại trùng hoàn toàn (có gắn cờ xung đột): {len(df_clean)} dòng")
print(f"- Sau khi áp quy tắc deduplicate hoàn toàn theo Product_ID: {len(df_clean_unique)} dòng")
```

Bài 8 Xử lý dữ liệu thiếu

```python
# 1. Báo cáo tỷ lệ thiếu sau khi chuyển kiểu
missing_report = pd.DataFrame({
    'Số lượng thiếu': df_clean.isna().sum(),
    'Tỷ lệ thiếu (%)': (df_clean.isna().mean() * 100).round(2)
}).sort_values(by='Số lượng thiếu', ascending=False)
print("=== BÁO CÁO DỮ LIỆU THIẾU TRƯỚC XỬ LÝ ===")
print(missing_report[missing_report['Số lượng thiếu'] > 0])

# 2. Price_VND: Điền median theo từng Category (chuẩn hơn median toàn bộ vì giá theo ngành hàng khác nhau rất lớn)
df_clean['Price_VND'] = df_clean.groupby('Category')['Price_VND'].transform(lambda x: x.fillna(x.median()))

# 3. Rating: Điền median theo Category & tạo cờ Rating_Missing_Flag
df_clean['Rating_Missing_Flag'] = df_clean['Rating'].isna().astype(int)
df_clean['Rating'] = df_clean.groupby('Category')['Rating'].transform(lambda x: x.fillna(x.median()))

# 4. Description: Điền nhãn và tạo cờ Description_Missing_Flag
df_clean['Description_Missing_Flag'] = df_clean['Description'].isna().astype(int)
df_clean['Description'] = df_clean['Description'].fillna("Không có mô tả").astype(str).str.strip()

# 5. Seller_City: Nếu còn thiếu thì gán 'Unknown' (tránh gán mode làm sai lệch phân bố địa lý)
df_clean['Seller_City'] = df_clean['Seller_City'].fillna("Unknown")

# 6. Created_Date: Không bịa ngày ngẫu nhiên; điền mốc median ngày của tập dữ liệu
median_date = df_clean['Created_Date'].dropna().median()
df_clean['Created_Date'] = df_clean['Created_Date'].fillna(median_date)

print("\nSố lượng thiếu còn lại sau Bài 8:\n", df_clean.isna().sum()[df_clean.isna().sum() > 0])
```

Bài 9 Xử lý ngoại lệ và giá trị ngoài miền

```python
# 1. Rating (Miền hợp lệ [1, 5]):
# Rating > 5 hoặc < 1 là lỗi nhập liệu -> gán median theo nhóm hàng
invalid_rating = (df_clean['Rating'] < 1) | (df_clean['Rating'] > 5)
category_rating_median = df_clean.loc[~invalid_rating].groupby('Category')['Rating'].median()
df_clean.loc[invalid_rating, 'Rating'] = df_clean.loc[invalid_rating, 'Category'].map(category_rating_median)

# 2. Price_VND (Lớn hơn 0):
# Giá âm hoặc <= 0 là lỗi -> thay bằng median nhóm
invalid_price = df_clean['Price_VND'] <= 0
category_price_median = df_clean.loc[~invalid_price].groupby('Category')['Price_VND'].median()
df_clean.loc[invalid_price, 'Price_VND'] = df_clean.loc[invalid_price, 'Category'].map(category_price_median)

# 3. Stock & Purchase_Count (Không âm):
# Số lượng âm là bất thường -> gán về 0
df_clean.loc[df_clean['Stock'] < 0, 'Stock'] = 0
df_clean.loc[df_clean['Purchase_Count'] < 0, 'Purchase_Count'] = 0

# 4. Stock = 9999 (Ngoại lệ tiềm năng):
# Gắn cờ để theo dõi, dùng IQR cắt ngưỡng trên (winsorization) để tránh lệch mô hình
Q1_stock = df_clean['Stock'].quantile(0.25)
Q3_stock = df_clean['Stock'].quantile(0.75)
IQR_stock = Q3_stock - Q1_stock
stock_upper_limit = Q3_stock + 1.5 * IQR_stock

df_clean['Stock_Outlier_Flag'] = (df_clean['Stock'] > stock_upper_limit).astype(int)
df_clean['Stock'] = np.where(df_clean['Stock'] > stock_upper_limit, stock_upper_limit, df_clean['Stock'])

# 5. Discount (Miền hợp lệ [0, 1]):
# Cắt ngưỡng clip về khoảng [0.0, 1.0]
df_clean['Discount'] = df_clean['Discount'].clip(lower=0.0, upper=1.0)

print("=== KIỂM TRA LẠI CÁC MIỀN GIÁ TRỊ ===")
print(f"Rating: [{df_clean['Rating'].min()}, {df_clean['Rating'].max()}]")
print(f"Price_VND: [{df_clean['Price_VND'].min():,.0f}, {df_clean['Price_VND'].max():,.0f}]")
print(f"Stock: [{df_clean['Stock'].min()}, {df_clean['Stock'].max()}]")
print(f"Discount: [{df_clean['Discount'].min()}, {df_clean['Discount'].max()}]")
```

BÀI 10: Tạo đặc trưng mới (Feature Engineering)

```python
# Ngày mốc tham chiếu theo đề bài: 01/09/2026
ref_date = pd.to_datetime('2026-09-01')

# 1. Tuổi của sản phẩm tính theo ngày
df_clean['Product_Age_Days'] = (ref_date - df_clean['Created_Date']).dt.days

# 2. Giá sau khi áp dụng chiết khấu
df_clean['Discounted_Price'] = df_clean['Price_VND'] * (1 - df_clean['Discount'])

# 3. Doanh thu ước tính
df_clean['Estimated_Revenue'] = df_clean['Discounted_Price'] * df_clean['Purchase_Count']

# 4. Độ dài phần mô tả sản phẩm (số ký tự)
df_clean['Description_Length'] = df_clean['Description'].astype(str).str.len()

print("=== 5 DÒNG ĐẦU CÁC ĐẶC TRƯNG MỚI TẠO ===")
print(df_clean[['Product_Age_Days', 'Discounted_Price', 'Estimated_Revenue', 'Description_Length']].head())
```

BÀI 13: TƯƠNG QUAN VÀ CHỌN ĐẶC TRƯNG

```python
# 58. Ma trận tương quan cho các biến số sau làm sạch
cot_so_mo_hinh = [
    "Price_VND", "Rating", "Purchase_Count", "Stock", "Weight", "Discount",
    "Product_Age_Days", "Discounted_Price", "Estimated_Revenue", "Description_Length"
]
ma_tran_tuong_quan = df_clean[cot_so_mo_hinh].corr()
print(ma_tran_tuong_quan.round(2))

# 59. Vẽ heatmap và tìm các cặp có |r| lớn
plt.figure(figsize=(9, 7))
sns.heatmap(ma_tran_tuong_quan, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            vmin=-1, vmax=1, square=True, cbar_kws={"shrink": 0.8})
plt.title("Ma trận tương quan các biến số (sau làm sạch)")
plt.tight_layout()
plt.show()

cap_tuong_quan = (
    ma_tran_tuong_quan
    .where(np.triu(np.ones(ma_tran_tuong_quan.shape), k=1).astype(bool))
    .stack()
    .rename("r")
    .reset_index()
    .rename(columns={"level_0": "Bien_1", "level_1": "Bien_2"})
)
cap_tuong_quan["abs_r"] = cap_tuong_quan["r"].abs()
cap_tuong_quan = cap_tuong_quan.sort_values("abs_r", ascending=False)

NGUONG_R = 0.5
cap_r_lon = cap_tuong_quan[cap_tuong_quan["abs_r"] >= NGUONG_R]
print(f"\nCác cặp biến có |r| >= {NGUONG_R}:")
print(cap_r_lon[["Bien_1", "Bien_2", "r"]])

# 60. (Câu hỏi lý thuyết — không cần code, trả lời trong Markdown/báo cáo:
#      tương quan cao không chứng minh nhân quả vì có thể do biến gây nhiễu,
#      quan hệ ngược chiều, hoặc do công thức tính toán như Estimated_Revenue.)

# 61. Loại cột định danh Product_ID khỏi tập đặc trưng mô hình
cot_dac_trung_mo_hinh = [c for c in df_encoded.columns
                          if c not in ["Product_ID", "Product_Name", "Description", "Target_High_Sales"]]
print("\nProduct_ID đã loại khỏi tập đặc trưng:", "Product_ID" not in cot_dac_trung_mo_hinh)
print("Số đặc trưng còn lại:", len(cot_dac_trung_mo_hinh))

# 62. VarianceThreshold và SelectKBest — nêu đặc trưng bị loại và lý do
from sklearn.feature_selection import VarianceThreshold, SelectKBest, f_classif

cot_so_cho_vt = [c for c in cot_dac_trung_mo_hinh if pd.api.types.is_numeric_dtype(df_encoded[c])]
X_vt = df_encoded[cot_so_cho_vt].fillna(0)

vt = VarianceThreshold(threshold=0.01)
vt.fit(X_vt)
dac_trung_bi_loai_vt = X_vt.columns[~vt.get_support()].tolist()
print("\nĐặc trưng bị loại bởi VarianceThreshold (phương sai quá thấp):")
print(dac_trung_bi_loai_vt)

X_kbest = df_encoded[cot_so_cho_vt].fillna(df_encoded[cot_so_cho_vt].median())
y_kbest = df_clean["Target_High_Sales"]

K = 8
skb = SelectKBest(score_func=f_classif, k=K)
skb.fit(X_kbest, y_kbest)

diem_so = pd.Series(skb.scores_, index=cot_so_cho_vt).sort_values(ascending=False)
dac_trung_bi_loai_kbest = diem_so.tail(len(diem_so) - K).index.tolist()

print(f"\nTop {K} đặc trưng theo SelectKBest:")
print(diem_so.head(K).round(2))
print("\nCác đặc trưng bị loại (điểm F-score thấp):")
print(dac_trung_bi_loai_kbest)
```
