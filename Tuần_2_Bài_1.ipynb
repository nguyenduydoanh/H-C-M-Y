Phần 2: Đọc dữ liệu từ file

```python
import pandas as pd
import numpy as np

# Đọc file từ thư mục 'học máy' trên Google Drive
df = pd.read_csv('/content/drive/MyDrive/bài 1 tuần 2/sample_data.csv')

# Xem trước 5 dòng đầu tiên
df.head()
```

Phần 3: Kiểm tra thông tin dữ liệu

```python
# Kiểm tra kích thước và kiểu dữ liệu
df.info()

# Thống kê mô tả
df.describe(include='all')
```

Phần 4: Kiểm tra dữ liệu thiếu

```python
# Đếm số lượng giá trị thiếu trên mỗi cột
df.isnull().sum()

# Tính tỉ lệ phần trăm thiếu
missing_ratio = df.isnull().mean() * 100
print(missing_ratio)
```

Phần 5: Xử lý dữ liệu thiếu

5.1. Xóa các hàng có giá trị thiếu

```python
df.dropna(inplace=True)
```

5.2. Hoặc điền giá trị mặc định

```python
# Điền giá trị trung bình cho cột 'Age'
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Điền giá trị thường gặp nhất cho cột 'Gender'
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

Phần 6: Xử lý dữ liệu trùng lặp

```python
# Kiểm tra bản ghi trùng
df.duplicated().sum()

# Xóa bản ghi trùng
df.drop_duplicates(inplace=True)
```

Phần 7: Chuẩn hóa dữ liệu

```python
# Chuẩn hóa chữ: ví dụ chuyển về chữ thường
df['Gender'] = df['Gender'].str.lower()
```

Phần 8: Chuyển đổi kiểu dữ liệu nếu cần

```python
# Chuyển cột 'Date' sang kiểu datetime
df['JoinDate'] = pd.to_datetime(df['JoinDate'])
```

Phần 9: Lưu lại dữ liệu đã làm sạch

```python
df.to_csv('sample_data_cleaned.csv', index=False)
```
